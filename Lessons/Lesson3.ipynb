How do we get information from a user?

The **input()** function allows user input.


Syntax is **input**(*prompt*) where prompt is a string message to the user

The input() function returns a string.

```python
# You can check the datatype of a variable using the type() function
f_name = input("Enter your first name: ")
l_name = input("Enter your Last name: ")
age = input("Enter your age: ")

print(f'{f_name} {l_name} is {age} years old.')
```

What if I need a number from a user to do some calculation?

We will need to convert the string to a number which is either and integer or float(decimial)



```python
sub_total = float( input("Enter the subtotal: ") )
taxes = float( input("Enter the taxes: ") )
total = sub_total + taxes
print(f' Subtotal: {sub_total} + taxes {taxes} = {total} Total Amt')
```

```python
price = float( input('enter the price of the item'))
taxes= price * 0.08
print(the taxes for your item of price {price} is {taxes}')
```

# Mathematical Operators
PEMDAS - order of operations

+ - adding
- - subtraction
* - multiplication
/ - division
** - Exponential function

```python
x=5
y=2
a=2*x+1
print(a)

a = x**3
print(a)

a = x/2
print(a)

a = x**(1/2)
print(a)

#integer division - where we divide 2 number and only get the integer portion
#of the division
print( x//y)
print(15*"#")
#Modulas operator (remander operator) -
print(x%y)
print(11%3)
print(0%3)

```

```python
#Mathematical Operator Shorthand
x=5
x+=1
print(x)
x-=1
print(x)
x*=2
print(x)
x/=2
print(x)
x%=2
print(x)
```

```python
#representing fraction (rational numbers)
a = 1/2
print(a)

a=1/3
print(a)

a=1/99
print(a)

a=1/7
print(a)
```
