# **Declaring variables in python**

* We used variables to store information that our programs will need.

* Python is completely object oriented

* variables are how store information

* Variables names should have meaningful names. Example:average, total_cost

* variables names must begin with a letter

* variables names are case sensitive. So average and Average are different variables

* variables names can contain upper and lower case letters and underscores '_'

* variables names made up of two or more words should be seperated by an underscore so that the code is easily readable.

* varibles names can not be keywords like: if, for, while, etc. Use: help("keywords") to get a list of python keywords

* Variable Datatypes is determined by python when the variable is assigned
